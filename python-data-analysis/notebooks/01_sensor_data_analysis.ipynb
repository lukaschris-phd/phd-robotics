```python
import numpy as np
import pandas as pd
import matplotlib as plt
```

```python
time = np.arrange(0, 10)

distance = np.array([150, 143, 135, 128, 120, 111, 103, 94, 86, 78])

plt.pot(time, distance)
plt.xlabel("Time (s)")
plt.ylabel("Distance (cm)")
plt.title("Robot Distance Sensor")
plt.show()
```
